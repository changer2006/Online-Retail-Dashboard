# Online Retail Data Analysis Project

## Project Goal

The objective of this project is to analyze retail transaction data and generate business insights related to:

- Sales Performance
- Customer Behavior
- Product Performance
- Customer Segmentation

This project will help answer important business questions such as:

- Which products generate the highest revenue?
- Which countries contribute the most sales?
- Who are the most valuable customers?
- How can customers be segmented for marketing strategies?

##### Step 6: Data Cleaning

`````````````python
df = df.drop_duplicates()
`````````````

`````````````python
df.isnull().sum()
`````````````

`````````````python
df = df.dropna(
    subset=["CustomerID"]
)
`````````````

`````````````python
df["Revenue"] = (
    df["Quantity"] *
    df["UnitPrice"]
)
`````````````
